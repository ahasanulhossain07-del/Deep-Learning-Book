## #Digit Character Class

```python
txt="the is 87 dollars"
x=re.findall(r"\d",txt)
print(x)
```

```python
txt="the is 87 dollars"
x=re.findall(r"\d+",txt)
print(x)
```

```python
txt="this is my number 01819-242334"
x=re.findall(r"\d{5}-\d{6}",txt)
print(x)
```

```python
# at begin
txt="149 is the 123 squre value"
x=re.findall(r"^\d+",txt)
print(x)
```

```python
txt = "149 is the square value 123"
x = re.findall(r"\d+$", txt)
print(x)
```
